```python
import torch
from torch import nn
from d2l import torch as d2l
import time
```

```python
def cpu(): #@save
    return torch.device("cpu")

def gpu(i=0):
    return torch.device(f"cuda:{i}")
```

```python
def num_gpus():
    return torch.cuda.device_count()
```

```python
#Transfering batch from CPU to GPU to perform computation on GPU.
@d2l.add_to_class(d2l.Trainer)
def __init__(self, max_epochs, num_gpus=0, gradient_clip_val=0):
    self.save_hyperparameters()
    self.gpus = [d2l.gpu(i) for i in range(min(num_gpus, d2l.num_gpus()))]


@d2l.add_to_class(d2l.Trainer)
def prepare_batch(self, batch):
    if self.gpus:
        batch = [a.to(device=self.gpus[0]) for a in batch]
    return batch

@d2l.add_to_class(d2l.Trainer)
def prepare_model(self, model):
    model.trainer = self
    model.board.xlim = [0, self.max_epochs]
    if self.gpus:
        model.to(devic=self.gpus[0])
    self.model = model

```

```python
def perform_matmul(size, show_result=True):
    start = time.time()
    results = []
    for i in range(1000):
        X = torch.randn(100, 100, device=gpu())
        Y = torch.randn(100, 100, device=gpu())
        Z = torch.matmul(X, Y)
        norm = torch.norm(Z, p="fro")
        results.append(norm)
        if show_result:
            print(norm)
    
    if show_result==False:
        for result in results:
            print(result)
    print("Time : ", time.time()-start)
```

```python
perform_matmul(100)
```
